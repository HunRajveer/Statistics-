```python
import numpy as np
from  scipy import stats
```

```python
# we are performing two-sample test using t-test and z-test respectively
```

```python
S1 = [11,22,33,44,55,66,77,88,99,100,120,432]
S2 = [12,23,34,45,56,67,78,89,90,87,65,66,97]
s1_mean = np.mean(S1)
s2_mean = np.mean(S2)
s1_std = np.std(S1)
s2_std = np.std(S2)
n1 = len(S1)
n2 = len(S2)
pop_std = 12
```

```python

# we have inbuild two sample test in our library
# we need to pass ttest and p-value at once because it will return tuple of (t-test , p-value)
t_test,p_value= stats.ttest_ind(S1,S2,equal_var=False)

```

```python
# Same approch as t-test and z-test for acceptence
alpha = 0.05
if p_value>alpha:
  print("accept null hypothesis")
else:
  print("reject null hypothesis")
```
