```python
import numpy as np
from scipy import stats
```

```python
s = [11,22,33,44,55,66,77,88,99,100,120,132]
s_mean= np.mean(s)
s_std = np.std(s)
n = len(s)
pop_mean = 80
```

```python
t_test = (s_mean-pop_mean)/(s_std/np.sqrt(n))
t_test
```

```python
p_value = 2*(1-stats.t.cdf(abs(t_test),df=n-1))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Accept Null Hypothesis")
```
