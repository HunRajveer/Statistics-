```python
import numpy as np
import seaborn as sns
```

```python
arr = [-43,-1,2,2,4,5,6,7,7,16,17,17,18,32,50,101]
arr = np.array(arr)
# we have an inbuild function to find quartiles known as percentile

```

```python
# find first quartile
q1 = np.percentile(arr,25)
q1
```

```python
#  Find third quartile
q3 = np.percentile(arr,75)
q3
```

```python
# IQR is interal quartile range
IQR = q3-q1
IQR
```

```python
# upperfance
UF = q3+(IQR*1.5)
UF
```

```python
LF = q1 - (IQR*1.5)
LF
```

```python
l = []
for i in arr:
  if i>=LF and i<=UF:
    l.append(i)
```

```python
arr2 = np.array(l)
arr2
```

```python
# box plot without outliers
sns.boxplot(y = arr2)
```

```python
# boxplot with outliers
sns.boxplot(y = arr)
```
