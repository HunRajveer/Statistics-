```python
import numpy as np
from scipy.stats import norm # this is the library where we use states function
```

```python
sample = [51,88,13,10,11,22,33,44,55,66,77,88,99,100,12,23,34,45,56,67,78,89,21,32,43,54,65,76,87,98,91]
# we  took sample size more then 30 to peform z-test (must check )

popstd = 40
popmean=70
sampmean=np.mean(sample)
n = len(sample)
```

```python
z_score = (sampmean-popmean)/(popstd/np.sqrt(n))
z_score
```

```python
pvalue = 2*(1-norm.cdf(abs(z_score))) #cdf is cumalitive distribution function
pvalue
```

```python
# final conclusion if pvalue is greater alpha (0.05) hear  then we accept it or else reject it
alpha = 0.05
if pvalue<alpha:
  print("reject  it")
else:
  print("accepted")
```
