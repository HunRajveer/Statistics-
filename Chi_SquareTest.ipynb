```python

import numpy as np
import seaborn as sns
import pandas as pd
from scipy import stats
```

```python
df=sns.load_dataset('titanic')
```

```python
df
```

```python
df['sex'],df['survived'].dropna()
```

```python
# our motive is to peform a chi square test in qualitative data (sex) and check wether it holds a relation with survived
```

```python
frame = pd.crosstab(df.survived,df.sex)
```

```python
frame
```

```python
# when we peform chi test it returns 4value in tuple forms so assign it we need to pass four variable
chi,p_value,dof,expected = stats.chi2_contingency(frame)
# chi2_contingency is the name for chi-square test
```

```python
chi
```

```python
dof
```

```python
p_value
# result np.float64(1.1973570627755645e-58) means power mai -58 very very small value
```

```python
expected
```

```python
# to give final-verdict we will compare p_value with signifence value
alpha=0.05
if p_value<alpha:
  print("null hypothises is rejected and there is a signifance relation between sex and survived")
else:
    print("there is no relationship between survervied and sex")
```
